# Pandas Tutorial: Learn Pandas in 30 Minutes

This notebook provides a comprehensive introduction to Pandas, the essential data manipulation and analysis library for Python.

**What is Pandas?**
- Pandas stands for **Panel Data** (from econometrics)
- Provides two main data structures: **Series** (1D) and **DataFrame** (2D)
- Built on top of NumPy for fast numerical operations
- Essential for data cleaning, exploration, transformation, and analysis
- Handles CSV, Excel, SQL, JSON, and many other data formats

```python
# Import Pandas - convention is to import as pd
import pandas as pd
import numpy as np

print(f"Pandas version: {pd.__version__}")
```

## 1. Series

A **Series** is a one-dimensional labeled array. Think of it as a single column of data with an index.

```python
# Creating a Series from a list
data = [100, 300, 490]
series = pd.Series(data, index=['a', 'b', 'c'])
print("Series with custom index:")
print(series)
print(f"\nType: {type(series)}")
```

```python
# Accessing elements
print(f"By label (loc):  series.loc['b']  = {series.loc['b']}")
print(f"By position (iloc): series.iloc[0] = {series.iloc[0]}")

# Modifying values
series.loc['b'] = 1293
print(f"\nAfter update: series.loc['b'] = {series.loc['b']}")

# Filtering
print(f"\nValues >= 500:")
print(series[series >= 500])
```

```python
# Creating a Series from a dictionary
s = pd.Series({'Alice': 25, 'Bob': 30, 'Charlie': 35})
print("Series from dict:")
print(s)
```

## 2. DataFrame - Creating and Basics

A **DataFrame** is a 2D labeled data structure with columns of potentially different types. Think of it as a spreadsheet or SQL table.

```python
# Creating a DataFrame from a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'London', 'Tokyo', 'Paris'],
    'Salary': [50000, 60000, 70000, 55000]
}

df = pd.DataFrame(data)
print("DataFrame:")
print(df)
```

```python
# Basic Attributes
print(f"Shape (rows, cols): {df.shape}")
print(f"\nColumn names: {df.columns.tolist()}")
print(f"\nData types:")
print(df.dtypes)
print(f"\nStatistical summary:")
print(df.describe())
```

## 3. Selecting Data

Pandas provides multiple ways to select columns, rows, and subsets of data.

```python
# ---- Select columns ----
print("Single column (Series):")
print(df['Name'])

print("\nMultiple columns (DataFrame):")
print(df[['Name', 'Salary']])
```

```python
# ---- Select rows ----
print("First row by position (iloc):")
print(df.iloc[0])

print("\nRows 1 to 2 by position:")
print(df.iloc[1:3])

print("\nFirst 2 rows (head):")
print(df.head(2))

print("\nLast 2 rows (tail):")
print(df.tail(2))
```

```python
# ---- loc vs iloc ----
# loc  -> label-based (uses index labels and column names)
# iloc -> integer-based (uses position numbers)

print("loc [0:2, 'Name':'City'] (inclusive on both ends):")
print(df.loc[0:2, 'Name':'City'])

print("\niloc [0:2, 0:2] (exclusive on end):")
print(df.iloc[0:2, 0:2])
```

## 4. Modifying DataFrames

Add, update, and remove columns and rows.

```python
# Add a new column
df['Bonus'] = df['Salary'] * 0.10
print("After adding Bonus column:")
print(df)

# Update a column
df['Salary'] = df['Salary'] + 5000
print("\nAfter salary raise:")
print(df)
```

```python
# Add a new row using concat
new_row = pd.DataFrame([{'Name': 'Fardin',
                         'Age': 21,
                         'City': 'Dhaka',
                         'Salary': 0,
                         'Bonus': 0}])

df = pd.concat([df, new_row], ignore_index=True)
print("After adding new row:")
print(df)
```

```python
# Drop a column
df_no_bonus = df.drop(columns=['Bonus'])
print("After dropping Bonus:")
print(df_no_bonus)

# Drop a row by index
df_no_last = df.drop(index=4)  # drop Fardin's row
print("\nAfter dropping last row:")
print(df_no_last)
```

## 5. Reading Files (CSV)

Pandas can read data from CSV, Excel, JSON, SQL, and many other formats.

```python
# Reading a CSV file
# df = pd.read_csv("data.csv")               # basic read
# df = pd.read_csv("data.csv", index_col="Name")  # set index column

# Other read functions:
# pd.read_excel("file.xlsx")
# pd.read_json("file.json")
# pd.read_sql(query, connection)
# pd.read_html("url")

# For this tutorial, let's create a sample DataFrame
np.random.seed(42)
pokemon_data = {
    'Type1': ['Grass', 'Fire', 'Water', 'Fire', 'Water', 'Grass', 'Electric', 'Water', 'Fire', 'Grass'],
    'Type2': ['Poison', None, None, 'Flying', 'Ice', None, None, 'Poison', 'Dragon', 'Flying'],
    'HP': [45, 39, 44, 78, 79, 80, 35, 65, 78, 65],
    'Height': [0.7, 0.6, 0.5, 1.7, 1.6, 2.0, 0.4, 0.8, 1.7, 2.0],
    'Weight': [6.9, 8.5, 9.0, 90.5, 85.5, 100.0, 6.0, 55.0, 110.5, 83.0],
    'Legendary': [False, False, False, False, False, False, False, False, True, False]
}
df = pd.DataFrame(pokemon_data, index=['Bulbasaur', 'Charmander', 'Squirtle', 'Charizard', 
                                        'Blastoise', 'Venusaur', 'Pikachu', 'Tentacruel',
                                        'Reshiram', 'Tropius'])
df.index.name = 'Name'

print("Pokemon DataFrame:")
print(df)
```

```python
# Selection by label (when index is set)
print("Single row by label:")
print(df.loc["Pikachu"])

print("\nSlice of rows with specific columns:")
print(df.loc["Charizard":"Blastoise", ["HP", "Height", "Weight"]])

print("\nBy position - every 3rd row, columns 1-3:")
print(df.iloc[1:8:3, 1:4])
```

## 6. Filtering Data

Filtering lets you select rows based on conditions - one of the most common operations in data analysis.

```python
# Filter by a condition
tall_pokemon = df[df["Height"] >= 2][["Type1", "Height"]]
print("Tall Pokemon (Height >= 2):")
print(tall_pokemon)

# Get names (index) of legendary pokemon as a list
legendary_pokemon = df[df["Legendary"] == True].index.tolist()
print(f"\nLegendary Pokemon: {legendary_pokemon}")
```

```python
# Multiple conditions with | (OR) and & (AND)
water_pokemon = df[(df["Type1"] == "Water") | (df["Type2"] == "Water")]
print("Water Pokemon (Type1 OR Type2 is Water):")
print(water_pokemon)

print("\nHeavy Fire Pokemon (Fire AND Weight > 50):")
heavy_fire = df[(df["Type1"] == "Fire") & (df["Weight"] > 50)]
print(heavy_fire)
```

```python
# Using isin() for multiple value checks
selected_types = df[df["Type1"].isin(["Fire", "Water"])]
print("Fire or Water Type1:")
print(selected_types)

# Using query() for cleaner syntax
result = df.query("HP > 60 and Height > 1.0")
print("\nHP > 60 and Height > 1.0 (using query):")
print(result)
```

## 7. Aggregation and Grouping

Aggregation functions summarize data, and `groupby` lets you apply them to groups within your data.

```python
# Basic aggregation
print("--- Basic Aggregation ---")
print(f"Mean of all numeric columns:")
print(df.mean(numeric_only=True))

print(f"\nSum of Height: {df['Height'].sum():.1f}")
print(f"Min HP: {df['HP'].min()}")
print(f"Max Weight: {df['Weight'].max()}")
print(f"Count (non-null): {df.count()['Type2']} (Type2 has nulls)")

print(f"\nMost common Weight values:")
print(df['Weight'].value_counts().head())
```

```python
# Grouping - split-apply-combine pattern
group = df.groupby("Type1")

print("Average Weight by Type1:")
print(group["Weight"].mean())

print("\nMultiple aggregations:")
print(group.agg({
    'HP': 'mean',
    'Weight': ['min', 'max'],
    'Height': 'mean'
}))
```

## 8. Sorting

Sort DataFrames by column values or by the index.

```python
# Sort by a column
print("Sorted by Weight (ascending):")
print(df.sort_values('Weight').head())

print("\nSorted by HP (descending):")
print(df.sort_values('HP', ascending=False).head())

# Sort by multiple columns
print("\nSorted by Type1 then Weight:")
print(df.sort_values(['Type1', 'Weight'], ascending=[True, False]))

# Sort by index
print("\nSorted by Name (index):")
print(df.sort_index().head())
```

## 9. Data Cleaning

Real-world data is messy. Pandas provides tools to handle missing values, duplicates, inconsistencies, and data type issues.

```python
# 1. Handle missing data
print("--- Missing Data ---")
print(f"Null counts per column:\n{df.isnull().sum()}")
print(f"\nType2 has {df['Type2'].isnull().sum()} missing values")

# Option A: Drop rows with missing values
df_dropped = df.dropna(subset=["Type2"])
print(f"\nAfter dropna: {len(df_dropped)} rows (was {len(df)})")

# Option B: Fill missing values
df_filled = df.fillna({"Type2": "None"})
print(f"\nAfter fillna:")
print(df_filled[['Type1', 'Type2']])
```

```python
# 2. Fix inconsistent values
df_clean = df.copy()

# Replace specific values
# df_clean["Type1"] = df_clean["Type1"].replace({"Grass": "Plant", "Water": "Aqua"})

# 3. Standardize text
print("Uppercase Type1:")
print(df_clean["Type1"].str.upper().head())

print("\nLowercase Type1:")
print(df_clean["Type1"].str.lower().head())

# 4. Fix data types
df_clean["Legendary"] = df_clean["Legendary"].astype(bool)
print(f"\nLegendary dtype: {df_clean['Legendary'].dtype}")

# 5. Remove duplicates
df_clean = df_clean.drop_duplicates()
print(f"After drop_duplicates: {len(df_clean)} rows")
```

## 10. String Operations

Pandas provides vectorized string methods via the `.str` accessor.

```python
names = pd.Series(['Alice Smith', 'Bob Johnson', 'Charlie Brown', 'Diana Prince'])

print(f"Original:\n{names}\n")
print(f"Upper: {names.str.upper().tolist()}")
print(f"Lower: {names.str.lower().tolist()}")
print(f"Length: {names.str.len().tolist()}")
print(f"Contains 'li': {names.str.contains('li').tolist()}")
print(f"First name: {names.str.split(' ').str[0].tolist()}")
print(f"Last name: {names.str.split(' ').str[-1].tolist()}")
print(f"Replace 'a' with '@': {names.str.replace('a', '@').tolist()}")
print(f"Starts with 'C': {names.str.startswith('C').tolist()}")
```

## 11. Merging and Joining DataFrames

Combine DataFrames using merge (SQL-like joins), concat (stacking), or join.

```python
# Two related DataFrames
employees = pd.DataFrame({
    'EmpID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'DeptID': [101, 102, 101, 103]
})

departments = pd.DataFrame({
    'DeptID': [101, 102, 103, 104],
    'DeptName': ['Engineering', 'Marketing', 'Sales', 'HR']
})

# Inner merge (only matching rows)
merged = pd.merge(employees, departments, on='DeptID', how='inner')
print("Inner merge:")
print(merged)

# Left merge (keep all from left)
print("\nLeft merge:")
print(pd.merge(employees, departments, on='DeptID', how='left'))

# Outer merge (keep all)
print("\nOuter merge:")
print(pd.merge(employees, departments, on='DeptID', how='outer'))
```

```python
# Concat - stacking DataFrames
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})

# Vertical concat (add rows)
print("Vertical concat:")
print(pd.concat([df1, df2], ignore_index=True))

# Horizontal concat (add columns)
print("\nHorizontal concat:")
print(pd.concat([df1, df2], axis=1))
```

## 12. Apply and Map

Apply custom functions to rows, columns, or individual elements.

```python
data = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Score': [85, 92, 78],
    'Salary': [50000, 60000, 45000]
})

# apply - works on a Series or DataFrame
data['Grade'] = data['Score'].apply(lambda x: 'A' if x >= 90 else 'B' if x >= 80 else 'C')
print("After apply (Grade):")
print(data)

# map - for element-wise mapping with a dict or function
grade_map = {'A': 'Excellent', 'B': 'Good', 'C': 'Average'}
data['Rating'] = data['Grade'].map(grade_map)
print("\nAfter map (Rating):")
print(data)

# apply on entire row (axis=1)
data['Summary'] = data.apply(lambda row: f"{row['Name']}: {row['Grade']}", axis=1)
print("\nAfter row-wise apply:")
print(data)
```

## 13. Dates and Time Series

Pandas has excellent support for dates, times, and time-series data.

```python
# Creating date ranges
d = pd.date_range('2023-01-01', periods=5, freq='D')   # Every day
w = pd.date_range('2023-01-01', periods=5, freq='W')   # Every week
m = pd.date_range('2023-01-01', periods=5, freq='ME')  # End of each month
h = pd.date_range('2023-01-01', periods=5, freq='h')   # Every hour

print(f"Daily:\n{d}\n")
print(f"Weekly:\n{w}\n")
print(f"Monthly:\n{m}\n")
print(f"Hourly:\n{h}")
```

```python
# Time series DataFrame
dates = pd.date_range('2023-01-01', periods=30, freq='D')
ts = pd.DataFrame({
    'Date': dates,
    'Sales': np.random.randint(100, 500, 30)
})
ts.set_index('Date', inplace=True)

print("Time Series:")
print(ts.head(10))

# Extract date components
print(f"\nYear : {ts.index.year[:3].tolist()}")
print(f"Month: {ts.index.month[:3].tolist()}")
print(f"Day  : {ts.index.day[:3].tolist()}")
print(f"Day of Week: {ts.index.dayofweek[:3].tolist()}  (0=Mon, 6=Sun)")

# Resample (like groupby for time series)
print(f"\nWeekly average sales:")
print(ts.resample('W').mean().head())
```

## 14. Pivot Tables

Pivot tables reshape and summarize data, similar to pivot tables in Excel.

```python
# Sample sales data
sales_data = pd.DataFrame({
    'Region': ['North', 'South', 'North', 'South', 'North', 'South'],
    'Product': ['A', 'A', 'B', 'B', 'A', 'B'],
    'Quarter': ['Q1', 'Q1', 'Q1', 'Q1', 'Q2', 'Q2'],
    'Revenue': [100, 150, 200, 120, 180, 90]
})

print("Sales Data:")
print(sales_data)

# Pivot table
pivot = pd.pivot_table(sales_data, values='Revenue', 
                       index='Region', columns='Product', 
                       aggfunc='sum')
print("\nPivot Table (Region x Product):")
print(pivot)

# Cross-tabulation
print("\nCross-tab (counts):")
print(pd.crosstab(sales_data['Region'], sales_data['Product']))
```

## Pandas Cheat Sheet

Quick reference for frequently used Pandas functions and operations.

### Creating DataFrames
| Function | Description |
|---|---|
| `pd.DataFrame(dict)` | From dictionary |
| `pd.DataFrame(list_of_dicts)` | From list of dicts |
| `pd.Series(data, index)` | Create a Series |
| `pd.read_csv('file.csv')` | Read CSV |
| `pd.read_excel('file.xlsx')` | Read Excel |
| `pd.read_json('file.json')` | Read JSON |

### Inspection
| Method | Description |
|---|---|
| `df.head(n)` / `df.tail(n)` | First / Last n rows |
| `df.shape` | (rows, columns) |
| `df.columns` | Column names |
| `df.dtypes` | Data types |
| `df.info()` | Summary (types, nulls) |
| `df.describe()` | Statistical summary |
| `df.to_string()` | Full print (no truncation) |

### Selection
| Syntax | Description |
|---|---|
| `df['col']` | Single column (Series) |
| `df[['col1', 'col2']]` | Multiple columns (DataFrame) |
| `df.loc[label]` | Row by label |
| `df.loc[r1:r2, c1:c2]` | Slice by labels (inclusive) |
| `df.iloc[pos]` | Row by position |
| `df.iloc[r1:r2, c1:c2]` | Slice by position (exclusive end) |

### Filtering
| Syntax | Description |
|---|---|
| `df[df['col'] > val]` | Filter by condition |
| `df[(cond1) & (cond2)]` | AND filter |
| `df[(cond1) \| (cond2)]` | OR filter |
| `df[df['col'].isin([...])]` | Filter by list of values |
| `df.query('col > val')` | Filter with query string |

### Modifying
| Method | Description |
|---|---|
| `df['new_col'] = values` | Add / Update column |
| `df.drop(columns=['col'])` | Drop column |
| `df.drop(index=i)` | Drop row |
| `pd.concat([df, new_row])` | Add row |
| `df.rename(columns={...})` | Rename columns |

### Aggregation
| Method | Description |
|---|---|
| `df.mean()` / `df.sum()` | Mean / Sum |
| `df.min()` / `df.max()` | Min / Max |
| `df.count()` | Non-null count |
| `df['col'].value_counts()` | Frequency of values |
| `df.describe()` | Summary statistics |

### Grouping
| Method | Description |
|---|---|
| `df.groupby('col')` | Group by column |
| `group['col'].mean()` | Group aggregation |
| `group.agg({...})` | Multiple aggregations |

### Sorting
| Method | Description |
|---|---|
| `df.sort_values('col')` | Sort by column |
| `df.sort_values('col', ascending=False)` | Sort descending |
| `df.sort_values(['c1','c2'])` | Sort by multiple columns |
| `df.sort_index()` | Sort by index |

### Data Cleaning
| Method | Description |
|---|---|
| `df.isnull()` / `df.notnull()` | Check for nulls |
| `df.dropna()` | Drop rows with nulls |
| `df.dropna(subset=['col'])` | Drop if specific col is null |
| `df.fillna(value)` | Fill nulls |
| `df.drop_duplicates()` | Remove duplicates |
| `df.replace({old: new})` | Replace values |
| `df['col'].astype(type)` | Change data type |

### String Methods (`.str`)
| Method | Description |
|---|---|
| `.str.upper()` / `.str.lower()` | Case conversion |
| `.str.len()` | String length |
| `.str.contains('x')` | Check contains |
| `.str.split(' ')` | Split strings |
| `.str.replace('a','b')` | Replace in strings |
| `.str.startswith('x')` | Check prefix |

### Merging & Joining
| Method | Description |
|---|---|
| `pd.merge(df1, df2, on='col')` | SQL-like join |
| `pd.merge(..., how='left')` | Left join |
| `pd.merge(..., how='outer')` | Full outer join |
| `pd.concat([df1, df2])` | Stack vertically |
| `pd.concat([df1, df2], axis=1)` | Stack horizontally |

### Apply & Map
| Method | Description |
|---|---|
| `df['col'].apply(func)` | Apply to each element |
| `df.apply(func, axis=1)` | Apply to each row |
| `df['col'].map(dict)` | Map values |

### Dates
| Function | Description |
|---|---|
| `pd.date_range(start, periods, freq)` | Create date range |
| `pd.to_datetime(col)` | Convert to datetime |
| `.dt.year` / `.dt.month` / `.dt.day` | Extract components |
| `df.resample('W').mean()` | Resample time series |
| Freq: `'D'` day, `'W'` week, `'ME'` month-end, `'h'` hour | Common frequencies |

### Exporting
| Method | Description |
|---|---|
| `df.to_csv('file.csv')` | Save as CSV |
| `df.to_excel('file.xlsx')` | Save as Excel |
| `df.to_json('file.json')` | Save as JSON |

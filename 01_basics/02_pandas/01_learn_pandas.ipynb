# Pandas Tutorial: Learn Pandas in 30 Minutes

This notebook provides a comprehensive introduction to Pandas, the essential data manipulation and analysis library for Python.

**What is Pandas?**
- Pandas stands for **Panel Data** (from econometrics)
- Provides two main data structures: **Series** (1D) and **DataFrame** (2D)
- Built on top of NumPy for fast numerical operations
- Essential for data cleaning, exploration, transformation, and analysis
- Handles CSV, Excel, SQL, JSON, and many other data formats


## 0. Imports


```python
# Import Pandas and NumPy
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

## 1. Series

A **Series** is a one-dimensional labeled array. Think of it as a single column of data with an index.


```python
# Creating a Series from a list
data = [100, 300, 490]
series = pd.Series(data, index=['a', 'b', 'c'])

print("Series:")
print(series)
print("\nType:", type(series))
```

```python
# Accessing elements
print("series.loc['b'] (access by label):", series.loc['b'])
print("series.iloc[0] (access by position):", series.iloc[0])
```

```python
# Modifying values
series.loc['b'] = 1293
series
```

```python
# Filtering
series[series >= 500]
```

```python
# Creating a Series from a dictionary
s = pd.Series({'Alice': 25, 'Bob': 30, 'Charlie': 35})
s
```

## 2. DataFrame - Creating and Basics

A **DataFrame** is a 2D labeled data structure with columns of potentially different types. Think of it as a spreadsheet or SQL table.


```python
# Creating a DataFrame from a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'London', 'Tokyo', 'Paris'],
    'Salary': [50000, 60000, 70000, 55000]
}
df = pd.DataFrame(data)
df
```

```python
# Basic attributes of DataFrame
print("Shape (rows, columns):", df.shape)
print("Columns:", df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nSummary statistics:")
print(df.describe())
```

## 3. Selecting Data

Pandas provides multiple ways to select columns, rows, and subsets of data.


```python
# Select columns
print("Single column (returns Series):")
print(df['Name'])

print("\nMultiple columns (returns DataFrame):")
print(df[['Name', 'Salary']])
```

```python
# Select rows by position
print("First row (iloc[0]):")
print(df.iloc[0])

print("\nRows 1-2 (iloc[1:3]):")
print(df.iloc[1:3])

print("\nFirst 2 rows (head(2)):")
print(df.head(2))

print("\nLast 2 rows (tail(2)):")
print(df.tail(2))
```

```python
# loc vs iloc comparison
# loc  -> label-based (uses index labels and column names)
# iloc -> integer-based (uses position numbers)

print("Using loc[0:2, 'Name':'City'] (label-based):")
print(df.loc[0:2, 'Name':'City'])

print("\nUsing iloc[0:2, 0:2] (position-based):")
print(df.iloc[0:2, 0:2])
```

## 4. Modifying DataFrames

Add, update, and remove columns and rows.


```python
# Add a new column
df['Bonus'] = df['Salary'] * 0.10
# Update a column
df['Salary'] = df['Salary'] + 5000
df
```

```python
# Add a new row using concat
new_row = pd.DataFrame([{'Name': 'Fardin',
                         'Age': 21,
                         'City': 'Dhaka',
                         'Salary': 0,
                         'Bonus': 0}])
df = pd.concat([df, new_row], ignore_index=True)
df
```

```python
# Drop a column
df_no_bonus = df.drop(columns=['Bonus'])
# Drop a row by index
df_no_last = df.drop(index=4)  # drop Fardin's row
(df_no_bonus, df_no_last)
```

## 5. Reading Files (CSV)

Pandas can read data from CSV, Excel, JSON, SQL, and many other formats.


```python
# Reading a CSV file
# df = pd.read_csv("data.csv")               # basic read
# df = pd.read_csv("data.csv", index_col="Name")  # set index column

# Other read functions:
# pd.read_excel("file.xlsx")
# pd.read_json("file.json")
# pd.read_sql(query, connection)
# pd.read_html("url")

# For this tutorial, let's create a sample DataFrame
np.random.seed(42)
pokemon_data = {
    'Type1': ['Grass', 'Fire', 'Water', 'Fire', 'Water', 'Grass', 'Electric', 'Water', 'Fire', 'Grass'],
    'Type2': ['Poison', None, None, 'Flying', 'Ice', None, None, 'Poison', 'Dragon', 'Flying'],
    'HP': [45, 39, 44, 78, 79, 80, 35, 65, 78, 65],
    'Height': [0.7, 0.6, 0.5, 1.7, 1.6, 2.0, 0.4, 0.8, 1.7, 2.0],
    'Weight': [6.9, 8.5, 9.0, 90.5, 85.5, 100.0, 6.0, 55.0, 110.5, 83.0],
    'Legendary': [False, False, False, False, False, False, False, False, True, False]
}
df = pd.DataFrame(pokemon_data, index=['Bulbasaur', 'Charmander', 'Squirtle', 'Charizard',
                                        'Blastoise', 'Venusaur', 'Pikachu', 'Tentacruel',
                                        'Reshiram', 'Tropius'])
df.index.name = 'Name'
df
```

```python
# Selection by label (when index is set)
print("Pikachu row:")
print(df.loc['Pikachu'])

print("\nCharizard-Blastoise rows, HP/Height/Weight columns:")
print(df.loc['Charizard':'Blastoise', ['HP', 'Height', 'Weight']])

print("\nEvery 3rd row from 1-8, columns 1-4:")
print(df.iloc[1:8:3, 1:4])
```

## 6. Filtering Data

Filtering lets you select rows based on conditions - one of the most common operations in data analysis.


```python
# Filter by a condition
print("Pokemon with Height >= 2:")
tall_pokemon = df[df['Height'] >= 2][['Type1', 'Height']]
print(tall_pokemon)

print("\nLegendary Pokemon:")
legendary_pokemon = df[df['Legendary'] == True].index.tolist()
print(legendary_pokemon)
```

```python
# Multiple conditions with | (OR) and & (AND)
# Always use parentheses around each condition

print("Water-type Pokemon (Type1 OR Type2 is Water):")
water_pokemon = df[(df['Type1'] == 'Water') | (df['Type2'] == 'Water')]
print(water_pokemon[['Type1', 'Type2', 'Weight']])

print("\nHeavy Fire-type Pokemon (Type1 is Fire AND Weight > 50):")
heavy_fire = df[(df['Type1'] == 'Fire') & (df['Weight'] > 50)]
print(heavy_fire[['Type1', 'Weight']])
```

```python
# Using isin() and query()
print("Pokemon that are Fire or Water type:")
selected_types = df[df['Type1'].isin(['Fire', 'Water'])]
print(selected_types[['Type1', 'HP']])

print("\nPokemon with HP > 60 and Height > 1.0:")
result = df.query('HP > 60 and Height > 1.0')
print(result[['Type1', 'HP', 'Height']])
```

## 7. Aggregation and Grouping

Aggregation functions summarize data, and `groupby` lets you apply them to groups within your data.


```python
# Basic aggregation
print("Mean values (numeric_only ignores text columns):")
print(df.mean(numeric_only=True))

print("\nTotal Height (sum):", df['Height'].sum())
print("Minimum HP (lowest):", df['HP'].min())
print("Maximum Weight (heaviest):", df['Weight'].max())
print("Missing Type2 values:", df['Type2'].isnull().sum())

print("\nWeight value counts (frequency of each weight):")
print(df['Weight'].value_counts().head())
```

```python
# Grouping - split-apply-combine pattern
# groupby splits data into groups, then applies operations per group

print("Mean Weight by Type:")
group = df.groupby('Type1')
print(group['Weight'].mean())

print("\nMultiple aggregations:")
print(group.agg({
     'HP': 'mean',              # Average HP per type
     'Weight': ['min', 'max'],  # Min and max weight per type
     'Height': 'mean'           # Average height per type
}))
```

## 8. Sorting

Sort DataFrames by column values or by the index.


```python
# Sorting examples
print("Sorted by Weight (ascending):")
print(df.sort_values('Weight').head()[['Type1', 'Weight']])

print("\nSorted by HP (descending):")
print(df.sort_values('HP', ascending=False).head()[['Type1', 'HP']])

print("\nSorted by Type1 (A-Z) then Weight (Z-A):")
print(df.sort_values(['Type1', 'Weight'], ascending=[True, False])[['Type1', 'Weight']])

print("\nSorted by index:")
print(df.sort_index().head())
```

## 9. Data Cleaning

Real-world data is messy. Pandas provides tools to handle missing values, duplicates, inconsistencies, and data type issues.


```python
# Have to add df.drop 
```

```python
# Handle missing data
print("Null counts per column:")
print(df.isnull().sum())

print("\nNull counts in Type2:", df['Type2'].isnull().sum())
print("Rows without null Type2:", df.dropna(subset=['Type2']).shape[0])

print("\nFilled Type2 nulls with 'None':")
print(df.fillna({'Type2': 'None'})[['Type1', 'Type2']].head())
```

```python
# Standardize text and fix data types
df_clean = df.copy()
df_clean['Legendary'] = df_clean['Legendary'].astype(bool)  # Convert to boolean
df_clean = df_clean.drop_duplicates()  # Remove duplicate rows

print("Type1 in uppercase:")
print(df_clean['Type1'].str.upper().head())

print("\nType1 in lowercase:")
print(df_clean['Type1'].str.lower().head())

print("\nLegendary data type:", df_clean['Legendary'].dtype)
print("Rows after removing duplicates:", len(df_clean))
```

## 10. String Operations

Pandas provides vectorized string methods via the `.str` accessor.


```python
names = pd.Series(['Alice Smith', 'Bob Johnson', 'Charlie Brown', 'Diana Prince'])

print("Original names:")
print(names.tolist())

print("\nUppercase (convert to UPPERCASE):")
print(names.str.upper().tolist())

print("\nLowercase (convert to lowercase):")
print(names.str.lower().tolist())

print("\nName lengths (count characters):")
print(names.str.len().tolist())

print("\nContains 'li' (boolean - True/False):")
print(names.str.contains('li').tolist())

print("\nFirst names (split by space, get [0]):")
print(names.str.split(' ').str[0].tolist())

print("\nLast names (split by space, get [-1]):")
print(names.str.split(' ').str[-1].tolist())

print("\nReplace 'a' with '@' (substitution):")
print(names.str.replace('a', '@').tolist())

print("\nStarts with 'C' (boolean check):")
print(names.str.startswith('C').tolist())
```

## 11. Merging and Joining DataFrames

Combine DataFrames using merge (SQL-like joins), concat (stacking), or join.


```python
# Two related DataFrames
employees = pd.DataFrame({
    'EmpID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'DeptID': [101, 102, 101, 103]
})
departments = pd.DataFrame({
    'DeptID': [101, 102, 103, 104],
    'DeptName': ['Engineering', 'Marketing', 'Sales', 'HR']
})

# how='inner' -> keep only matching keys (intersection)
print("Inner merge (only matching DeptIDs):")
print(pd.merge(employees, departments, on='DeptID', how='inner'))

# how='left' -> keep all rows from left DataFrame
print("\nLeft merge (keep all employees):")
print(pd.merge(employees, departments, on='DeptID', how='left'))

# how='outer' -> keep all rows from both (union)
print("\nOuter merge (keep all rows):")
print(pd.merge(employees, departments, on='DeptID', how='outer'))
```

```python
# Concat - stacking DataFrames
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})

print("Vertical concat (stack rows):")
print(pd.concat([df1, df2], ignore_index=True))

print("\nHorizontal concat (stack columns):")
print(pd.concat([df1, df2], axis=1))
```

## 12. Apply and Map

Apply custom functions to rows, columns, or individual elements.


```python
data = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Score': [85, 92, 78],
    'Salary': [50000, 60000, 45000]
})
data['Grade'] = data['Score'].apply(lambda x: 'A' if x >= 90 else 'B' if x >= 80 else 'C')
grade_map = {'A': 'Excellent', 'B': 'Good', 'C': 'Average'}
data['Rating'] = data['Grade'].map(grade_map)
data['Summary'] = data.apply(lambda row: f"{row['Name']}: {row['Grade']}", axis=1)
data
```

## 13. Dates and Time Series

Pandas has excellent support for dates, times, and time-series data.


```python
# Creating date ranges
# Frequency codes: D=Day, W=Week, ME=Month End, h=Hour

print("Daily frequency (5 days):")
d = pd.date_range('2023-01-01', periods=5, freq='D')
print(d)

print("\nWeekly frequency (5 weeks):")
w = pd.date_range('2023-01-01', periods=5, freq='W')
print(w)

print("\nEnd of month frequency (5 months):")
m = pd.date_range('2023-01-01', periods=5, freq='ME')
print(m)

print("\nHourly frequency (5 hours):")
h = pd.date_range('2023-01-01', periods=5, freq='h')
print(h)
```

```python
# Time series DataFrame
dates = pd.date_range('2023-01-01', periods=30, freq='D')
ts = pd.DataFrame({
    'Date': dates,
    'Sales': np.random.randint(100, 500, 30)
})
ts.set_index('Date', inplace=True)  # Use Date as the index

print("First 10 rows of time series:")
print(ts.head(10))

print("\nTime components (extract year, month, day):")
print("Years:", ts.index.year[:3].tolist())
print("Months:", ts.index.month[:3].tolist())
print("Days:", ts.index.day[:3].tolist())
print("Day of week (0=Monday, 6=Sunday):", ts.index.dayofweek[:3].tolist())

print("\nResampled to weekly with mean (group by week, average):")
print(ts.resample('W').mean().head())
```

## 14. Pivot Tables

Pivot tables reshape and summarize data, similar to pivot tables in Excel.


```python
# Sample sales data
sales_data = pd.DataFrame({
    'Region': ['North', 'South', 'North', 'South', 'North', 'South'],
    'Product': ['A', 'A', 'B', 'B', 'A', 'B'],
    'Quarter': ['Q1', 'Q1', 'Q1', 'Q1', 'Q2', 'Q2'],
    'Revenue': [100, 150, 200, 120, 180, 90]
})

print("Original sales data:")
print(sales_data)

print("\nPivot table (sum by Region x Product):")
pivot = pd.pivot_table(sales_data, values='Revenue',
                       index='Region', columns='Product',
                       aggfunc='sum')
print(pivot)

print("\nCrosstab (count by Region x Product):")
print(pd.crosstab(sales_data['Region'], sales_data['Product']))
```

## Pandas Cheat Sheet

Quick reference for the patterns shown above.

### Creation & IO
| Function | Description |
|---|---|
| `pd.Series(data, index=...)` | Create a 1D labeled Series |
| `pd.DataFrame(dict)` | Create a 2D DataFrame from a dict of columns |
| `pd.read_csv(path)` | Load a CSV file |
| `pd.read_excel(path)` | Load an Excel file |
| `pd.read_json(path)` | Load a JSON file |
| `pd.read_sql(query, conn)` | Run a SQL query |
| `df.to_csv(path, index=False)` | Write to CSV |
| `df.to_excel(path, index=False)` | Write to Excel |
| `df.to_json(path, orient='records')` | Write to JSON |

### Inspection & Selection
| Function | Description |
|---|---|
| `df.head(n)`, `df.tail(n)` | First / last `n` rows |
| `df.shape` | (rows, columns) |
| `df.dtypes` | Data type per column |
| `df.info()` | Dtypes + non-null counts |
| `df.describe()` | Summary statistics |
| `df['col']`, `df[['c1','c2']]` | Select column(s) |
| `df.loc[row, col]` | Label-based access |
| `df.iloc[i, j]` | Integer-position access |
| `df[mask]` | Boolean filter |
| `df.query(expr)` | Filter using a query string |
| `df.isin(values)` | Membership filter |

### Cleaning
| Function | Description |
|---|---|
| `df.isnull()`, `df.notnull()` | Null mask |
| `df.dropna(subset=[cols])` | Drop rows with nulls |
| `df.fillna(value)` | Replace nulls |
| `df.drop_duplicates()` | Remove duplicate rows |
| `df.astype(dtype)` | Cast column types |
| `df.replace({...})` | Replace specific values |
| `df['col'].str.upper()` | Vectorized string ops |

### Aggregation & Grouping
| Function | Description |
|---|---|
| `df.mean()`, `df.sum()`, `df.min()`, `df.max()` | Per-column aggregates |
| `df['col'].value_counts()` | Frequency of each value |
| `df.groupby(col).agg({...})` | Aggregate per group |
| `df.pivot_table(values, index, columns, aggfunc)` | Pivot table |
| `pd.crosstab(a, b)` | Cross-tabulation |

### Sorting & Reshaping
| Function | Description |
|---|---|
| `df.sort_values(col, ascending=...)` | Sort by column(s) |
| `df.sort_index()` | Sort by row index |
| `pd.concat([df1, df2])` | Stack / concatenate |
| `pd.merge(left, right, on=..., how=...)` | SQL-style join |
| `df.melt()`, `df.pivot()` | Long <-> wide reshape |

### Apply & Time Series
| Function | Description |
|---|---|
| `df['col'].apply(func)` | Element-wise function on a Series |
| `df.apply(func, axis=1)` | Row-wise function on a DataFrame |
| `df['col'].map(dict)` | Map values via dictionary |
| `pd.date_range(start, periods, freq)` | Date index |
| `df.resample(rule).mean()` | Resample by time bucket |

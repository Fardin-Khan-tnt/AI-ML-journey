# Step 5: Ask a Question about the data
- Try to answer a question you have about the data using a plot or statistic.

What are the locations with the fastest roller coasters (minimum of 10)?

```python
ax = df.query('Location != "Other"') \
    .groupby('Location')['Speed_mph'] \
    .agg(['mean','count']) \
    .query('count >= 10') \
    .sort_values('mean')['mean'] \
    .plot(kind='barh', figsize=(12, 5), title='Average Coast Speed by Location')
ax.set_xlabel('Average Coaster Speed')
plt.show()
```

## EDA Cheat Sheet

Quick reference for the patterns shown above.

### Data Inspection
| Command | Description |
|---|---|
| `df.shape` | (rows, columns) |
| `df.head(n)` / `df.tail(n)` | First/last n rows |
| `df.columns` | Column index |
| `df.dtypes` | Per-column dtypes |
| `df.describe()` | Summary stats for numeric columns |
| `df.info()` | dtypes + non-null counts + memory |
| `df.isna().sum()` | Missing values per column |
| `df.duplicated()` / `df.duplicated(subset=[...])` | Boolean duplicate mask |

### Selection and Filtering
| Command | Description |
|---|---|
| `df[['col_a', 'col_b']]` | Select columns |
| `df.loc[mask]` / `df.iloc[rows]` | Label / integer based selection |
| `df.query('col > 5')` | SQL-like filtering |
| `df.drop(['col'], axis=1)` | Drop columns |
| `df.dropna()` / `df.fillna(v)` | Handle missing values |

### Cleaning and Transformation
| Command | Description |
|---|---|
| `df.rename(columns={...})` | Rename columns |
| `pd.to_datetime(df['date'])` | Parse dates |
| `df['col'].astype(int)` | Cast dtype |
| `df.duplicated(subset=[...])` then `.reset_index(drop=True)` | Deduplicate |

### Univariate Analysis
| Command | Description |
|---|---|
| `df['col'].value_counts()` | Frequency of each category |
| `df['col'].plot(kind='bar')` | Bar chart of counts |
| `df['col'].plot(kind='hist', bins=n)` | Histogram |
| `df['col'].plot(kind='kde')` | Density estimate |
| `df['col'].plot(kind='box')` | Box plot |

### Feature Relationships
| Command | Description |
|---|---|
| `df.plot(kind='scatter', x=, y=)` | Scatter plot |
| `sns.scatterplot(x=, y=, hue=, data=df)` | Scatter with grouping |
| `sns.pairplot(df, vars=[...], hue=...)` | Pairwise scatter grid |
| `df.corr()` | Pearson correlation matrix |
| `sns.heatmap(df.corr(), annot=True)` | Correlation heatmap |

### Grouping and Aggregation
| Command | Description |
|---|---|
| `df.groupby('k')['v'].mean()` | Mean per group |
| `df.groupby('k')['v'].agg(['mean','count'])` | Multiple aggregations |
| `df.groupby('k')['v'].size()` | Group sizes |
| `.sort_values('mean')` / `.sort_values('mean', ascending=False)` | Sort results |

### Asking Questions of the Data
| Pattern | Description |
|---|---|
| `df.query('cond').groupby('k')['v'].agg(['mean','count']).query('count >= N').sort_values('mean')` | Filter, group, aggregate, threshold, sort |
| `.plot(kind='barh')` after `sort_values` | Horizontal bar chart for ranked results |
